# Data Preprocessing

```python
samples = samples.drop('depth (ft)', axis=1)\
    .set_index('Sample')
print(samples.shape)
samples
```

```python
sample_pressures = sample_pressures.rename(
    columns={
        col: f"r_{col}"
        for col in sample_pressures.columns
        if any(char.isdigit() for char in str(col))  # Ensure col is converted to string
    }
)
print(sample_pressures.shape)
sample_pressures.T
```

```python
sample_pressures = sample_pressures.T.reset_index()
sample_pressures = sample_pressures.drop('index', axis=1)
sample_pressures.index.name = 'Sample'
print(sample_pressures.shape)
sample_pressures
```

```python
df = samples.join(sample_pressures).sort_index(ascending=True)
print(df.shape)
df
```

# Calcs

```python
import math
INTERFACIAL_TENSION = 484
CONTACT_ANGLE_DEGREES = 140

def degree_to_radian(degrees:float):
    return degrees * math.pi / 180

def J(
        capillary_pressure:float,
        wetting_phase_saturation:float,
        permeability:float,
        porosity:float,
        interfacial_tension:float = INTERFACIAL_TENSION,
        contact_angle_degrees:float = CONTACT_ANGLE_DEGREES,
):
    numerator = capillary_pressure * wetting_phase_saturation
    denominator = interfacial_tension * math.cos(degree_to_radian(CONTACT_ANGLE_DEGREES))
    if denominator == 0 or porosity == 0:
        raise ValueError("denominator or porosity cannot be zero")
    quotient = numerator / denominator
    scaling_factor = math.sqrt(permeability/porosity)
    return quotient * scaling_factor

for sample in df.index:
    try:
        df.loc[sample, 'J'] = J(
            capillary_pressure=df.loc[sample, 'pc'],
            wetting_phase_saturation=df.loc[sample, 'wp'],
            permeability=df.loc[sample, 'permeability (md)'],
            porosity=df.loc[sample, 'porosity (%)'],
        )
    except ValueError as e:
        print(f"Error in sample {sample}: {e}")
```
