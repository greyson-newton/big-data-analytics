```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define a signal that we can apply convolution to
x = [4.2, 4.3, 4.4, 4.1, 4.2, 4.3, 4.5, 4.1,4.4, 4.2,4.3,4.2,4.4 ,10.1, 10.3, 10.4, 10.5, 10.3, 9.8, 9.9, 9.8, 2.2, 2.4, 2.3, 2.1]
```

```python
# Inputs (1d signal)
fig,ax = plt.subplots()
ax.plot(x, 'k-')
plt.show()
```

```python
# A zero-padded convolution operation (kernel size=3, stride=1, dilation=1)
def conv_1d(x_in, omega):
    index = 0
    x_out = np.zeros_like(x_in)
    for count in x_in:
      if index == 0:
      # starting convolution to implement the first zero
        x_out[index] = omega[1]*x_in[index]+omega[2]*x_in[index+1]
      if (0<index)  and  (index<len(x_in)-1):
      # middle convolution to apply all the weights
        x_out[index] = omega[0]*x_in[index-1]+omega[1]*x_in[index]+omega[2]*x_in[index+1]
      if index == len(x_in)-1:
      # ending convolution to implement the last zero
        x_out[index] = omega[0]*x_in[index-1]+omega[1]*x_in[index]
      index += 1
    print (omega)
    return x_out
```

```python
omega = [0.33,0.33,0.33]
h1 = conv_1d(x, omega)
fig,ax = plt.subplots()
ax.plot(x, 'k-',label='before')
ax.plot(h1, 'r-',label='after')
ax.legend()
plt.show()
```
